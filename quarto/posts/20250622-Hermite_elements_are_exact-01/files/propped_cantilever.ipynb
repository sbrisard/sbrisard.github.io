# The propped cantilever

```python
import sympy as sp
```

```python
x, L, p, EI, a, b = sp.symbols(r"x L p EI a b")
```

## Exact solution

```python
v = p * x**4 / 24 / EI + a * x**3 + b * x**2
v
```

```python
assert EI * v.diff(x, 4) == p
```

Boundary condition at $x = L$: $v(L) = 0$.

```python
eq_b = v.subs(x, L)
eq_b
sol_b = sp.solve(eq_b, b)
v = v.subs(b, sol_b[0]).expand()
v
```

Boundary condition at $x = L$: $M(L) = 0$.

```python
M = sp.expand(EI * v.diff(x, 2))

eq_a = M.subs(x, L)
sol_a = sp.solve(eq_a, a)[0]
sp.Eq(a, sol_a)
```

```python
v = v.subs(a, sol_a).factor()
v
```

```python
v.diff(x).subs(x, L)
```

```python
M = sp.expand(EI * v.diff(x, 2))
M.subs(x, 0)
```

## Approximate solution

```python
v_approx = a * x**2 * (x - L)
v_approx
```

```python
θ_approx = v_approx.diff(x)
κ_approx = θ_approx.diff(x)

assert v_approx.subs(x, 0) == 0
assert θ_approx.subs(x, 0) == 0
assert v_approx.subs(x, L) == 0
```

```python
U = sp.integrate(EI * κ_approx**2 / 2, (x, 0, L))
U
```

```python
V = sp.integrate(p * v_approx, (x, 0, L))
V
```

```python
Π = U - V
Π
```

```python
eq_a_bis = Π.diff(a)
sol_a_bis = sp.solve(eq_a_bis, a)[0]
sp.Eq(a, sol_a_bis)
```

```python
v_approx = v_approx.subs(a, sol_a_bis).factor()
v_approx.factor()
```

```python
M_approx = sp.expand(EI * v_approx.diff(x, 2))
M_approx.subs(x, 0)
```

```python
v_approx.diff(x).subs(x, L)
```
